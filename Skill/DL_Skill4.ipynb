```python
import torch

if torch.cuda.is_available():
    device = "cuda"
else:
    device = "cpu"

a = torch.tensor([10, 20, 30], device=device)
b = a * 2

print("Device:", device)
print("Input Tensor:", a)
print("Result:", b)
```

```python
import subprocess

result = subprocess.run(
    ["pip", "freeze"],
    capture_output=True,
    text=True
)

with open("environment_info.txt", "w") as f:
    f.write("Python Version: " + sys.version + "\n")
    f.write("PyTorch Version: " + torch.__version__ + "\n")
    f.write("CUDA Version: " + str(torch.version.cuda) + "\n")
    f.write("cuDNN Version: " + str(torch.backends.cudnn.version()) + "\n")
    f.write("\nInstalled Packages:\n")
    f.write(result.stdout)

print("Environment information saved successfully!")
```
